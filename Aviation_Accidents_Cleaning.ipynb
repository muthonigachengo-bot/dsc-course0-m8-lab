## Data Cleaning

### Filtering aircrafts and events

We want to filter the dataset to include aircraft that the client is interested in an analysis of:
- inspect relevant columns
- figure out any reasonable imputations
- filter the dataset

```python
# Inspect columns relevant for filtering
filter_columns = [
    "Event.Date",
    "Aircraft.Category",
    "Amateur.Built",
    "Make",
    "Model"
]

display(aviation_df[filter_columns].head())

# Check for missing values in the relevant columns
print("Missing values:")
display(aviation_df[filter_columns].isna().sum())

# Inspect the unique values that will be used for filtering
for column in filter_columns:
    print(f"\n{column}")
    display(aviation_df[column].value_counts(dropna=False).head(20))


# Reasonable imputations
# Convert Event.Date to datetime
aviation_df["Event.Date"] = pd.to_datetime(
    aviation_df["Event.Date"],
    errors="coerce"
)

# Drop rows missing essential filtering information
aviation_df = aviation_df.dropna(
    subset=["Event.Date", "Aircraft.Category", "Amateur.Built", "Make", "Model"]
)

# Filter the dataset
# Keep accidents from 1983 onwards
aviation_df = aviation_df[
    aviation_df["Event.Date"].dt.year >= 1983
]

# Keep only professionally built aircraft
aviation_df = aviation_df[
    aviation_df["Amateur.Built"] == "No"
]

# Keep only airplanes
aviation_df = aviation_df[
    aviation_df["Aircraft.Category"] == "Airplane"
]

print("Filtered dataset shape:", aviation_df.shape)
display(aviation_df.head())
```

### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
# Clean injury-related columns
injury_columns = [
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

# Convert injury columns to numeric
for col in injury_columns:
    aviation_df[col] = pd.to_numeric(aviation_df[col], errors="coerce")

# Assume missing injury counts indicate zero reported injuries
aviation_df[injury_columns] = aviation_df[injury_columns].fillna(0)

# Construct key injury metrics
# Estimate total occupants (crew + passengers)
aviation_df["Total.Occupants"] = (
    aviation_df["Total.Fatal.Injuries"] +
    aviation_df["Total.Serious.Injuries"] +
    aviation_df["Total.Minor.Injuries"] +
    aviation_df["Total.Uninjured"]
)

# Total fatal or serious injuries
aviation_df["Fatal.Serious.Injuries"] = (
    aviation_df["Total.Fatal.Injuries"] +
    aviation_df["Total.Serious.Injuries"]
)

# Proportion of occupants experiencing fatal or serious injuries
aviation_df["Fatal.Serious.Rate"] = (
    aviation_df["Fatal.Serious.Injuries"] /
    aviation_df["Total.Occupants"]
)

# Handle division by zero
aviation_df["Fatal.Serious.Rate"] = (
    aviation_df["Fatal.Serious.Rate"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)
# Inspect the derived metrics
display(
    aviation_df[
        [
            "Total.Occupants",
            "Fatal.Serious.Injuries",
            "Fatal.Serious.Rate"
        ]
    ].head()
)
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python

# Clean Aircraft.damage column
# Inspect existing values
display(aviation_df["Aircraft.damage"].value_counts(dropna=False))

# Standardize text formatting
aviation_df["Aircraft.damage"] = (
    aviation_df["Aircraft.damage"]
    .astype(str)
    .str.strip()
    .str.title()
    )
# Replace missing values with "Unknown"
aviation_df["Aircraft.damage"] = aviation_df["Aircraft.damage"].replace(
    {"Nan": "Unknown"}
)


# Construct destruction metric
# Binary indicator:
# 1 = Aircraft destroyed
# 0 = Aircraft not destroyed
aviation_df["Destroyed"] = (
    aviation_df["Aircraft.damage"] == "Destroyed"
).astype(int)

# Verify the cleaned results
display(aviation_df["Aircraft.damage"].value_counts(dropna=False))
display(aviation_df["Destroyed"].value_counts())
```

### Cleaning the Make Column

The `Make` column identifies the aircraft manufacturer. Before analysis, the following cleaning tasks were performed:

- Standardized manufacturer names by trimming whitespace and applying consistent title-case formatting.

- Replaced placeholder values such as "Unknown" and "Nan" with missing values (NaN).

- Removed records with missing manufacturer names because they cannot be attributed to a specific aircraft make.

- To ensure statistically robust comparisons, only manufacturers with at least 50 accident records were retained for analysis. This reduces the risk of drawing conclusions from very small sample sizes.

```python
# ---------------------------------------
# Investigate and clean the Make column
# ---------------------------------------

# Inspect the most common make values
display(aviation_df["Make"].value_counts(dropna=False).head(30))

# Check missing values
print("Missing Make values:", aviation_df["Make"].isna().sum())

# ---------------------------------------
# Cleaning tasks
# ---------------------------------------
# 1. Convert to string
# 2. Remove leading/trailing whitespace
# 3. Standardize capitalization
# 4. Replace obvious missing placeholders
# 5. Remove rows with missing/unknown make values

aviation_df["Make"] = (
    aviation_df["Make"]
    .astype(str)
    .str.strip()
    .str.title()
)

# Replace common placeholders with NaN
aviation_df["Make"] = aviation_df["Make"].replace({
    "": np.nan,
    "Unknown": np.nan,
    "Nan": np.nan
})

# Drop rows where Make is missing
aviation_df = aviation_df.dropna(subset=["Make"])

# ---------------------------------------
# Keep only makes with a reasonable sample size
# ---------------------------------------

make_counts = aviation_df["Make"].value_counts()

# Threshold for statistical robustness
min_make_count = 50

valid_makes = make_counts[make_counts >= min_make_count].index

aviation_df = aviation_df[
    aviation_df["Make"].isin(valid_makes)
]
# ---------------------------------------
# Verify results
# ---------------------------------------

print(f"Number of makes retained: {len(valid_makes)}")
print(f"Dataset shape after make filtering: {aviation_df.shape}")

display(aviation_df["Make"].value_counts().head(20))
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python
# Clean the Model column

# Remove rows with missing model information
aviation_df = aviation_df.dropna(subset=["Model"])

# Standardize model names
aviation_df["Model"] = (
    aviation_df["Model"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Inspect the most common models
display(aviation_df["Model"].value_counts().head(20))


# Check whether model names are unique to a make
# Count the number of makes associated with each model
model_make_counts = (
    aviation_df.groupby("Model")["Make"]
    .nunique()
    .sort_values(ascending=False)
)

print("Models associated with more than one make:")
display(model_make_counts[model_make_counts > 1].head(20))


# Create a unique aircraft identifier

aviation_df["Aircraft.Type"] = (
    aviation_df["Make"] + " " + aviation_df["Model"]
)

# Inspect the new identifier
display(
    aviation_df[
        ["Make", "Model", "Aircraft.Type"]
    ].head()
)
```
